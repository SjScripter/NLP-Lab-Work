# Experiment-4: Term Frequency and NER

**4.1. To perform Term-Frequency Analysis and Named Entity Recognition on a given text using any toolkit.**

●	Calculate the frequency of each word and save the output in CSV format (Term, Frequency)

●	Identify the top 10 most frequent terms and display their corresponding frequencies.


```python
!pip install -q spacy
!python -m spacy download en_core_web_sm

import spacy
import pandas as pd
import string
from collections import Counter

nlp = spacy.load("en_core_web_sm")

with open("4.1_4.2_input.txt", "r", encoding="utf-8") as f:
    text = f.read()

doc = nlp(text)

# Term Frequency
words = [
    token.text.lower()
    for token in doc
    if not token.is_stop and not token.is_punct and token.is_alpha
]

freq = Counter(words)

df = pd.DataFrame(freq.items(), columns=["Term", "Frequency"])
df = df.sort_values(by="Frequency", ascending=False)

df.to_csv("term_frequency.csv", index=False)

print("Top 10 Frequent Terms")
print(df.head(10))

print("\nNamed Entities")
for ent in doc.ents:
    print(ent.text, "->", ent.label_)
```

**4.2. To perform Term-Frequency Analysis on a given text without using any toolkit.**

●	Calculate the frequency of each word and save the output in CSV format (Term, Frequency)

●	Identify the top 10 most frequent terms and display their corresponding frequencies.


```python
import string
import pandas as pd
from collections import Counter

with open("4.1_4.2_input.txt", "r", encoding="utf-8") as f:
    text = f.read()

text = text.lower()

translator = str.maketrans("", "", string.punctuation)
text = text.translate(translator)

words = text.split()

freq = Counter(words)

df = pd.DataFrame(freq.items(), columns=["Term", "Frequency"])
df = df.sort_values(by="Frequency", ascending=False)

df.to_csv("term_frequency_without_toolkit.csv", index=False)

print("Top 10 Frequent Terms")
print(df.head(10))
```

# 4.3. Write a Python program without using NLTK, spaCy, or any other NLP toolkit to:

1.	Read multiple text documents.
2.	Convert text to lowercase.
3.	Remove punctuation.
4.	Tokenize the documents.
5.	Calculate TF for every term.
6.	Calculate DF for every term.
7.	Calculate IDF using the logarithmic formula.
8.	Calculate TF-IDF for every term in every document.
9.	Display Term, TF, DF, IDF, and TF-IDF at each step.
10.	Identify the top 10 terms with the highest TF-IDF score for each document.


```python
import math
import string
from collections import Counter

documents = [
    "Natural language processing is a field of artificial intelligence.",
    "Natural language processing helps computers understand human language.",
    "Machine learning is an important part of artificial intelligence."
]

processed_docs = []

translator = str.maketrans("", "", string.punctuation)

for doc in documents:
    doc = doc.lower()
    doc = doc.translate(translator)
    tokens = doc.split()
    processed_docs.append(tokens)

N = len(processed_docs)

# ---------------- TF ----------------

tf_list = []

for tokens in processed_docs:
    total = len(tokens)
    count = Counter(tokens)

    tf = {}
    for word in count:
        tf[word] = count[word] / total

    tf_list.append(tf)

# ---------------- DF ----------------

df = {}

for tokens in processed_docs:
    unique = set(tokens)
    for word in unique:
        df[word] = df.get(word, 0) + 1

# ---------------- IDF ----------------

idf = {}

for word in df:
    idf[word] = math.log(N / df[word])

# ---------------- TF-IDF ----------------

tfidf_docs = []

for tf in tf_list:
    scores = {}
    for word in tf:
        scores[word] = tf[word] * idf[word]
    tfidf_docs.append(scores)

# ---------------- Display ----------------

for i in range(N):

    print("=" * 60)
    print(f"Document {i+1}")
    print("=" * 60)

    print("{:<18}{:<10}{:<10}{:<12}{:<12}".format(
        "Term", "TF", "DF", "IDF", "TF-IDF"))

    for word in sorted(tfidf_docs[i]):
        print("{:<18}{:<10.4f}{:<10}{:<12.4f}{:<12.4f}".format(
            word,
            tf_list[i][word],
            df[word],
            idf[word],
            tfidf_docs[i][word]
        ))

    print("\nTop 10 TF-IDF Terms")

    top = sorted(tfidf_docs[i].items(),
                 key=lambda x: x[1],
                 reverse=True)[:10]

    for word, score in top:
        print(f"{word:20} {score:.4f}")

    print()
```
