# **Experiment 2.1:  Write a Python program to perform basic text preprocessing on a given data. The program should:**

Convert text to lowercase. (Also display how many uppercase?)

Remove punctuation.        (Also display how many different types of punctuations?)

Remove numbers.            (Also display how many numbers?)

Remove extra whitespace. (How much extra whitespace is present?)

Display the cleaned text.    



```python
import os
print(os.listdir())
```

```python
with open("2.1_text_data.txt", "r", encoding="utf-8") as file:
    original_text = file.read()
print("Original Text:")
print(original_text)
```

```python
import string
import re

text = original_text

#1
uppercase_count = sum(ch.isupper() for ch in text)
text = text.lower()

# 2.
punctuations = [ch for ch in text if ch in string.punctuation]
punctuation_count = len(punctuations)
different_punctuations = set(punctuations)

text = text.translate(str.maketrans("", "", string.punctuation))

# 3.
number_count = sum(ch.isdigit() for ch in text)
text = re.sub(r'\d+', '', text)

# 4.
extra_spaces = re.findall(r' {2,}|\t+', text)
extra_whitespace = sum(len(x) - 1 for x in extra_spaces)
text = re.sub(r'\s+', ' ', text).strip()

#results
print(":::Answers are as follows:::")

print("\nuppercase letters:", uppercase_count)

print("\npunctuation marks:", punctuation_count)
print("Different punctuation:", len(different_punctuations))
print("Punctuation types:", sorted(different_punctuations))
print("\nTotal numbers:", number_count)
print("\nWhitespaces:", extra_whitespace)
print("\n::Final Clean Text is As follows::")
print(text)
```

# Experiment 2.2: Write a Python program to perform word tokenization and sentence tokenization on a given text. Display the resulting words and sentences separately.

           (Three separate code: using NLTK, using spaCy and without using NLTK/spacy)


# Method 1 : Using NLTK

```python
with open("2.2_tokenization_data.txt", "r", encoding="utf-8") as file:
    text = file.read()
print(text)
```

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')

from nltk.tokenize import word_tokenize, sent_tokenize
sentences = sent_tokenize(text)
words = word_tokenize(text)
print("SENTENCES:")
for sentence in sentences:
    print(sentence)
print("\nWORDS:")
print(words)
```

```python
import re
sentences = re.split(r'(?<=[.!?])\s+', text.strip())

words = re.findall(r'\b\w+\b', text)
print("SENTENCES:")
for sentence in sentences:
    print(sentence)

print("\nWORDS:")
print(words)
```

# Experiment 2.3: Write a Python program to tokenize a text and remove commonly occurring stop words such as is, the, a, an, in, of, and, to, etc. Display the original and filtered tokens. (Also, display Stop Words Separately).

```python
with open("2.3_clean_data.txt", "r", encoding="utf-8") as file:
    text = file.read()

print("Original Text:")
print(text)
```

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

tokens = word_tokenize(text)

stop_words = set(stopwords.words('english'))

found_stop_words = [word for word in tokens  if word.lower() in stop_words]

filtered_tokens = [word for word in tokens if word.lower() not in stop_words]

print("ORIGINAL TOKENS")
print(tokens)

print("\nSTOP WORDS")
print(found_stop_words)

print("\nFILTERED TOKENS")
print(filtered_tokens)
```
