# **Experiment-1: Data Analysis**

**Experiment 1.1: Read the employee dataset and display the average salary of each department using a bar chart.**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv("employee_information_100.csv")

avg_salary_df = data.groupby('Department')['Salary'].mean().sort_values(ascending=False).reset_index()

plt.figure(figsize=(8, 5))
bars = plt.bar(
    avg_salary_df['Department'],
    avg_salary_df['Salary'],
    color=['#1f77b4', '#aec7e8', '#ffbb78'],
    edgecolor='black'
)

plt.ylim(0, max(avg_salary_df['Salary']) * 1.15)
plt.title('Average Salary by Department', fontsize=14, fontweight='bold')
plt.xlabel('Department', fontsize=12)
plt.ylabel('Average Salary (INR)', fontsize=12)
plt.bar_label(bars, fmt='%d INR', fontsize=10, fontweight='bold')
plt.tight_layout()
plt.show()
```

**Experiment 1.2: Display the number of employees working in each department.**

```python
dept_count = data.groupby("Department")["Employee_ID"].count().reset_index()
dept_count.columns = ["Department", "Employee_Count"]

print(dept_count)
```

**Experiment 1.3: Display the percentage of male and female employees using a pie chart.**

```python
gender_count = data['Gender'].value_counts()

plt.figure(figsize=(8, 8))
plt.pie(
    gender_count,
    labels=gender_count.index,
    autopct="%1.1f%%",
    colors=['#ff9999', '#66b3ff'],
    startangle=90
)

plt.title("Employee Gender Distribution")
plt.show()
```

**Experiment 1.4: Draw a histogram showing salary distribution.**

```python
plt.figure(figsize=(8, 8))
plt.hist(
    data['Salary'],
    bins=5,
    color='#4caf50',
    edgecolor='black',
    alpha=0.7
)

plt.title('Employee Salary Distribution', fontsize=14, fontweight='bold')
plt.xlabel('Salary (INR)', fontsize=12)
plt.ylabel('Number of Employees', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

**Experiment 1.5: Draw a scatter plot showing the relationship between experience and salary.**

```python
plt.figure(figsize=(12, 10))
plt.scatter(
    data['Experience_Years'],
    data['Salary'],
    color='#1f77b4',
    s=100,
    alpha=0.8,
    edgecolor='grey',
    marker='o'
)

plt.title("Relationship between Experience and Salary", fontsize=14, fontweight='bold')
plt.xlabel('Years of Experience', fontsize=12)
plt.ylabel('Salary (INR)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

**Experiment 1.6: Display the top 10 employees with the highest salary.**

```python
top_10_highest = data.nlargest(10, 'Salary')
print(top_10_highest)
```

**Experiment 1.7: Find the highest salary in every department.**

```python
highest_salaries = data.groupby('Department')['Salary'].max().reset_index()
print(highest_salaries)
```

**Experiment 1.8: Display employees whose salary is greater than the overall average salary.**

```python
average_salary = data['Salary'].mean()
print(f"Overall Average Salary: {average_salary}")

above_average_sal = data[data['Salary'] > average_salary]
print("Employees Above Average Salary:\n")
print(above_average_sal)
```

**Experiment 1.9: Calculate the average years of experience in each department.**

```python
average_year_exper = data.groupby('Department')['Experience_Years'].mean().reset_index()
average_year_exper.columns = ['Department', 'Average_Years_Experience']

print(average_year_exper)
```

**Experiment 1.10: Display the distribution of employee ages using a histogram.**

```python
plt.figure(figsize=(6, 6))
plt.hist(
    data['Age'],
    bins=5,
    color='skyblue',
    edgecolor='black'
)

plt.title("Distribution of Employee Age", fontsize=14)
plt.xlabel('Age Groups', fontsize=12)
plt.ylabel('Number of Employees', fontsize=12)
plt.grid(axis='y', alpha=0.75)
plt.show()
```
