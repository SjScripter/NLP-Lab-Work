**Experiment-5: Subword Level Tokenization and POS Tagging**

```python
# BPE Tokenization
from tokenizers import Tokenizer
from tokenizers.models import BPE
from tokenizers.trainers import BpeTrainer
from tokenizers.pre_tokenizers import Whitespace

# Pretrained BPE
from transformers import AutoTokenizer

# SentencePiece
import sentencepiece as spm

# POS Tagging
import spacy
import nltk

# Frequency
from collections import Counter

#NLTK
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
# Open the input data file
file_path = "input_sub_word_data.txt"

with open(file_path, "r", encoding="utf-8") as file:
    text = file.read()

print("Input Data:\n")
print(text)
```

Experiment 5.1: Implement subword-level tokenization using a BPE-based tokenizer-
* Tokenize the sentence into subword units.
* Display the generated tokens and their corresponding token IDs.

(Implement two separate codes: a. Using pretrained model and b. Without using pretrained model)

```python
#5.1(a) Using pretrained model:

# Load pretrained BPE tokenizer
bpe_tokenizer = AutoTokenizer.from_pretrained("gpt2")

# Tokenize the input text
tokens = bpe_tokenizer.tokenize(text)

# Convert tokens into token IDs
token_ids = bpe_tokenizer.convert_tokens_to_ids(tokens)

# Display tokens and token IDs
print("BPE Tokenization using Pretrained Model\n")
print(f"{'Token':30} {'Token ID'}")
print("-" * 45)

for token, token_id in zip(tokens, token_ids):
    print(f"{token:30} {token_id}")
```

```python
#5.1(b) Without using pretrained model:

# Create BPE tokenizer
bpe = Tokenizer(BPE(unk_token="[UNK]"))

# Use whitespace pre-tokenization
bpe.pre_tokenizer = Whitespace()

# Define BPE trainer
trainer = BpeTrainer(
    vocab_size=500,
    min_frequency=2,
    special_tokens=["[UNK]", "[PAD]", "[CLS]", "[SEP]", "[MASK]"]
)

# Train BPE tokenizer using our input data
bpe.train(
    files=[file_path],
    trainer=trainer
)

# Tokenize input text
output = bpe.encode(text)

# Display tokens and token IDs
print("BPE Tokenization without Pretrained Model\n")
print(f"{'Token':30} {'Token ID'}")
print("-" * 45)

for token, token_id in zip(output.tokens, output.ids):
    print(f"{token:30} {token_id}")

print("\nVocabulary Size:", bpe.get_vocab_size())
```

Experiment 5.2: Implement subword-level tokenization using a SentencePiece tokenizer-

Tokenize the sentence into subword units.
Display the generated tokens and their corresponding token IDs.
(Implement two separate codes: a. Using pretrained model and b. Without using pretrained model)

```python
#5.2(a) Using pretrained model

# Load pretrained SentencePiece-based tokenizer
sp_tokenizer = AutoTokenizer.from_pretrained(
    "google/mt5-small",
    use_fast=False
)

# Tokenize the input text
sp_tokens = sp_tokenizer.tokenize(text)

# Convert tokens into token IDs
sp_token_ids = sp_tokenizer.convert_tokens_to_ids(sp_tokens)

# Display tokens and token IDs
print("SentencePiece Tokenization using Pretrained Model\n")
print(f"{'Token':30} {'Token ID'}")
print("-" * 45)

for token, token_id in zip(sp_tokens, sp_token_ids):
    print(f"{token:30} {token_id}")
```

```python
# 5.2(b) SentencePiece Tokenization Without Pretrained Model

# Train our own SentencePiece tokenizer
spm.SentencePieceTrainer.train(
    input=file_path,
    model_prefix="my_sentencepiece",
    vocab_size=300,
    model_type="bpe",
    character_coverage=1.0
)

# Load the trained SentencePiece model
sp = spm.SentencePieceProcessor(
    model_file="my_sentencepiece.model"
)

# Generate subword tokens
sp_tokens = sp.encode(text, out_type=str)

# Generate token IDs
sp_token_ids = sp.encode(text, out_type=int)

# Display tokens and token IDs
print("SentencePiece Tokenization without Pretrained Model\n")
print(f"{'Token':30} {'Token ID'}")
print("-" * 45)

for token, token_id in zip(sp_tokens, sp_token_ids):
    print(f"{token:30} {token_id}")

print("\nVocabulary Size:", sp.get_piece_size())
```

Exeriment 5.3: Implement Part-of-Speech (POS) tagging on a given text and
display the unique token, POS tag, and description of each POS tag-

"The young student is reading an interesting book in the library."   
 (Implement two separate codes: a. Using Spacy and b.  NLTK )


```python
pos_text = "The young student is reading an interesting book in the library."
```

```python
#5.2(a) Using SpaCy

# Load English spaCy model
nlp = spacy.load("en_core_web_sm")

# Process the sentence
doc = nlp(pos_text)

# Display POS information
print("POS Tagging using spaCy\n")
print(f"{'Token':20} {'POS Tag':10} {'Description'}")
print("-" * 60)

for token in doc:
    print(
        f"{token.text:20} "
        f"{token.pos_:10} "
        f"{spacy.explain(token.pos_)}"
    )
```

```python
#5.3(b) Using NLTK

# Tokenize sentence
nltk_tokens = nltk.word_tokenize(pos_text)

# Perform POS tagging
nltk_tags = nltk.pos_tag(nltk_tokens)

# Display POS information
print("POS Tagging using NLTK\n")
print(f"{'Token':20} {'POS Tag'}")
print("-" * 40)

for token, tag in nltk_tags:
    print(f"{token:20} {tag}")
```

Experiment 5.4: Implement Part-of-Speech (POS) tagging on a given text and display the unique token, POS tag, description of each token and frequency using Spacy

```python
# Process the input data using spaCy
doc = nlp(text)

# Calculate frequency of each token
frequency = Counter(
    token.text.lower()
    for token in doc
    if not token.is_space
)

# Store unique token information
token_info = {}

for token in doc:

    if token.is_space:
        continue

    word = token.text.lower()

    if word not in token_info:
        token_info[word] = {
            "POS": token.pos_,
            "Description": spacy.explain(token.pos_),
            "Frequency": frequency[word]
        }

# Display result
print("POS Tagging with Frequency using spaCy\n")
print(
    f"{'Token':20} "
    f"{'POS Tag':10} "
    f"{'Description':20} "
    f"{'Frequency'}"
)

print("-" * 75)

for word, info in token_info.items():

    print(
        f"{word:20} "
        f"{info['POS']:10} "
        f"{str(info['Description']):20} "
        f"{info['Frequency']}"
    )
```
